```python
%%writefile app.py

import streamlit as st

st.set_page_config(
    page_title="My Business App",
    page_icon="📦"
)

st.title("📦 My Business App")
st.write("Welcome to our business!")

menu = st.sidebar.selectbox(
    "Menu",
    ["Home", "Products", "Contact"]
)

if menu == "Home":
    st.header("🏠 Home")
    st.write("Welcome to our business!")
    st.info("Quality products at reasonable prices.")

elif menu == "Products":
    st.header("📦 Our Products")

    product = st.selectbox(
        "Select Product",
        ["Product 1 - ₹500", "Product 2 - ₹800", "Product 3 - ₹1200"]
    )

    quantity = st.number_input(
        "Quantity",
        min_value=1,
        max_value=10,
        value=1
    )

    if product == "Product 1 - ₹500":
        price = 500
    elif product == "Product 2 - ₹800":
        price = 800
    else:
        price = 1200

    total = price * quantity

    st.write(f"### 💰 Total Amount: ₹{total}")

    customer_name = st.text_input("Customer Name")

    if st.button("🛒 Place Order"):
        if customer_name:
            st.success("Order placed successfully! 🎉")
            st.write(f"Customer: {customer_name}")
            st.write(f"Product: {product}")
            st.write(f"Quantity: {quantity}")
            st.write(f"Total: ₹{total}")
        else:
            st.warning("Please enter your name.")

elif menu == "Contact":
    st.header("📞 Contact Us")
    st.write("📱 Phone: YOUR_PHONE_NUMBER")
    st.write("📧 Email: YOUR_EMAIL")
    st.write("📍 Address: YOUR_ADDRESS")
```

```python
%%writefile app.py

import streamlit as st

st.set_page_config(
    page_title="My Business App",
    page_icon="📦"
)

st.title("📦 My Business App")
st.write("Welcome to our business!")

menu = st.sidebar.selectbox(
    "Menu",
    ["Home", "Products", "Contact"]
)

if menu == "Home":
    st.header("🏠 Home")
    st.write("Welcome to our business!")
    st.info("Quality products at reasonable prices.")

elif menu == "Products":
    st.header("🛍️ Our Products")

    product = st.selectbox(
        "Select Product",
        ["Product 1 - ₹500", "Product 2 - ₹800", "Product 3 - ₹1200"]
    )

    # Display description and stock for the selected product
    if product == "Product 1 - ₹500":
        st.write("📝 Description: Premium quality product")
        st.write("📦 Stock: 20")

    elif product == "Product 2 - ₹800":
        st.write("📝 Description: High quality business product")
        st.write("📦 Stock: 15")

    else:
        st.write("📝 Description: Premium special product")
        st.write("📦 Stock: 10")

    quantity = st.number_input(
        "Quantity",
        min_value=1,
        max_value=10,
        value=1
    )

    # Calculate price based on selected product
    if product == "Product 1 - ₹500":
        price = 500
    elif product == "Product 2 - ₹800":
        price = 800
    else:
        price = 1200

    total = price * quantity

    st.success(f"💰 Total Amount: ₹{total}")

    customer_name = st.text_input("👤 Customer Name")

    if st.button("🛒 Place Order"):
        if customer_name:
            st.success("🎉 Order placed successfully!")

            st.write(f"👤 Customer: {customer_name}")
            st.write(f"📦 Product: {product}")
            st.write(f"🔢 Quantity: {quantity}")
            st.write(f"💰 Total: ₹{total}")
        else:
            st.warning("⚠️ Please enter your name.")

elif menu == "Contact":
    st.header("📞 Contact Us")
    st.write("📱 Phone: 9876543210")
    st.write("📧 Email: mybusiness@gmail.com")
    st.write("📍 Address: Khalilabad, Uttar Pradesh")
    st.info("For orders or enquiries, please contact us.")
```
